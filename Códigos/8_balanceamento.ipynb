## RandomUnderSampling

```python
from imblearn.under_sampling import RandomUnderSampler
undersample = RandomUnderSampler(sampling_strategy='not minority', random_state=42)
X_resampled, y_resampled = undersample.fit_resample(X_treino, y_treino)
```

```python
np.savez('../Bases/Criados/Treino_Teste/random.npz', X_treino=X_resampled, X_teste=X_teste, y_treino=y_resampled, y_teste=y_teste, feature_names=X_treino.columns)
```

```python
df_random = pd.concat([pd.DataFrame(X_resampled, columns=X_treino.columns),
                       pd.DataFrame(y_resampled, columns=['Doença_cardíaca'])], axis=1)
df_random.to_csv("../Bases/Criados/Treino_Teste/random.csv", index=False)
```

## Cross-validation

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import KFold, cross_val_score

X = X_resampled
y = y_resampled

clf = DecisionTreeClassifier(random_state=42)

k_folds = KFold(n_splits = 10, shuffle=True, random_state=42)

scores = cross_val_score(clf, X, y, cv = k_folds)

print("Cross Validation Scores: ", scores)
print("Average CV Score: ", scores.mean())
print("Number of CV Scores used in Average: ", len(scores))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import KFold, cross_val_score

clf = RandomForestClassifier(random_state=42)
k_folds = KFold(n_splits=10, shuffle=True, random_state=42)
scores = cross_val_score(clf, X_resampled, y_resampled, cv=k_folds)

print("Cross Validation Scores: ", scores)
print("Average CV Score: ", scores.mean())
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB

modelos = {
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    'Random Forest': RandomForestClassifier(random_state=42),
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42),
    'SVM': SVC(random_state=42),
    'KNN': KNeighborsClassifier(),
    'Naive Bayes': GaussianNB()
}

from sklearn.model_selection import KFold, cross_validate

scoring = {
    'accuracy': 'accuracy',
    'f1': 'f1',
    'recall': 'recall'
}

k_folds = KFold(n_splits=10, shuffle=True, random_state=42)

resultados = {}

for nome, modelo in modelos.items():
    resultado_cv = cross_validate(modelo, X_resampled, y_resampled, cv=k_folds, scoring=scoring)
    resultados[nome] = resultado_cv
    print(f"\nModelo: {nome}")
    for metric in scoring.keys():
        scores = resultado_cv[f'test_{metric}']
        print(f"{metric.capitalize()}: Média = {scores.mean():.4f} | Desvio = {scores.std():.4f}")

```

## Árvore de Decisão

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.tree import DecisionTreeClassifier, plot_tree
from scipy.stats import randint
from sklearn.model_selection import RandomizedSearchCV
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Parâmetros para busca aleatória
param_dist = {
    'criterion': ['gini', 'entropy'],
    'max_depth': randint(1, 50),
    'min_samples_split': randint(2, 20),
    'min_samples_leaf': randint(1, 10),
}

# Função para plotar a árvore
def plot_decision_tree(model, feature_names, class_names, title):
    plt.figure(figsize=(30, 15))
    plot_tree(model, filled=True, feature_names=feature_names, class_names=class_names, rounded=True, fontsize=10)
    plt.title(title)
    plt.show()

# RandomizedSearchCV com 100 iterações e 5-fold CV
random_search = RandomizedSearchCV(
    DecisionTreeClassifier(random_state=42),
    param_dist,
    n_iter=100,
    cv=5,
    scoring='accuracy',
    random_state=42
)

# Treinar nos dados balanceados
random_search.fit(X_resampled, y_resampled)

# Mostrar os melhores hiperparâmetros encontrados
print("\nMelhores parâmetros encontrados com RandomizedSearchCV:")
print(random_search.best_params_)
print("Melhor média de acurácia na validação cruzada: {:.2f}".format(random_search.best_score_))
```

```python
best_model_tree = DecisionTreeClassifier(
    criterion='entropy',
    max_depth=6,
    min_samples_leaf=3,
    min_samples_split=5,
    random_state=42
)

# Treinar o modelo nos dados balanceados
best_model_tree.fit(X_resampled, y_resampled)

plot_decision_tree(
    best_model_tree,
    feature_names=X_resampled.columns,
    class_names=['Sem DC', 'Com DC'],
    title="Árvore de Decisão - Parâmetros Fixos"
)
```

## Resultados

```python
previsoes_tree = best_model_tree.predict(X_teste)
accuracy_tree = accuracy_score(y_teste, previsoes_tree)
print("Acurácia do melhor modelo (RandomizedSearchCV) no conjunto de teste: {:.2f}".format(accuracy_tree))
```

```python
print(classification_report(y_teste, previsoes_tree))
```

## Floresta Aleatória

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from skopt import BayesSearchCV
from skopt.space import Integer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

param_bayes = {
    'criterion':  ['gini', 'entropy'],
    'max_depth': Integer(1, 50),
    'min_samples_split': Integer(2, 20),
    'min_samples_leaf': Integer(1, 10),
    'n_estimators': Integer(50, 120),
    'max_features': Integer(2, 5)
}

bayes_search = BayesSearchCV(RandomForestClassifier(), param_bayes, n_iter=50, cv=5, scoring='accuracy', random_state=42)

bayes_search.fit(X_resampled, y_resampled)

print("\nMelhores parâmetros encontrados com BayesSearchCV:")
print(bayes_search.best_params_)
print("Melhor pontuação de validação cruzada: {:.2f}".format(bayes_search.best_score_))
```

```python
best_model_bayes = RandomForestClassifier(
    criterion='entropy',
    max_depth=50,
    max_features=2,
    min_samples_leaf=6,
    min_samples_split=2,
    n_estimators=120,
    random_state=42
)

# Treinar nos dados balanceados
best_model_bayes.fit(X_resampled, y_resampled)

# Fazer previsões no conjunto de teste
previsoes_teste = best_model_bayes.predict(X_teste)

print(classification_report(y_teste, previsoes_teste))

cfm_matrix_teste = confusion_matrix(y_teste, previsoes_teste)
sns.heatmap(cfm_matrix_teste, cmap='coolwarm', annot=True, linewidth=1, fmt='d')
plt.title("Matriz de Confusão - Teste")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Obter a importância das features do modelo treinado
importancias = best_model_bayes.feature_importances_

# Usar os nomes das features correspondentes ao treino balanceado
nomes_das_features = X_treino.columns  

# Ordenar índices das features da maior para menor importância
indices_ordenados = np.argsort(importancias)[::-1]

# Exibir as features mais importantes com proteção contra erro de índice
print("Variáveis mais relevantes:")
for idx in indices_ordenados:
    if idx < len(nomes_das_features):
        print(f"{nomes_das_features[idx]}: {importancias[idx]:.4f}")
    else:
        print(f"[AVISO] Índice {idx} fora do range dos nomes das features")

# Plotar a importância das features
plt.figure(figsize=(10, 6))
plt.title("Importância das Features")

plt.barh(range(len(importancias)), importancias[indices_ordenados], align="center")

# Converter nomes para array e proteger o ytick labels
nomes_array = np.array(nomes_das_features)
nomes_array_safe = nomes_array[indices_ordenados[:len(nomes_array)]]

plt.yticks(range(len(nomes_array_safe)), nomes_array_safe)
plt.gca().invert_yaxis()  # para as mais importantes ficarem no topo
plt.tight_layout()
plt.show()
```
