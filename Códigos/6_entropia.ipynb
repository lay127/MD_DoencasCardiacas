```python
import pandas as pd
from scipy.stats import entropy
import numpy as np

def calcular_entropia_normalizada_por_coluna(df):
    entropias_normalizadas = {}
    for coluna in df.columns:
        # Frequências relativas (exclui NaN)
        frequencias = df[coluna].value_counts(normalize=True, dropna=True)
        n = len(frequencias)  # número de categorias distintas

        if n > 1:
            h = entropy(frequencias, base=2)
            h_max = np.log2(n)
            entropias_normalizadas[coluna] = h / h_max  # normalizada entre 0 e 1
        else:
            entropias_normalizadas[coluna] = 0.0  # sem incerteza se só tem um valor

    return entropias_normalizadas

df = pd.read_csv("../Bases/Criados/5_base_codificada.csv")
entropias_norm = calcular_entropia_normalizada_por_coluna(df)
for coluna, valor in entropias_norm.items():
    print(f"{coluna}: {valor:.4f}")
```
