## RandomUnderSampling

```python
X_prev = base.drop(columns=['Doença_cardíaca']) #variáveis independentes -> antes do balanceamento
y_classe = base['Doença_cardíaca'] #variável dependente -> antes do balanceamento
```

```python
from imblearn.under_sampling import RandomUnderSampler
undersample = RandomUnderSampler(sampling_strategy='not minority', random_state=42)
X_resampled, y_resampled = undersample.fit_resample(X_prev, y_classe)
```

## Cross-validation

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.model_selection import cross_val_score

#dados
X = X_treino
y = y_treino

#modelo
clf = DecisionTreeClassifier(random_state=42)

#validação cruzada
k_folds = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)

#execução da validação cruzada
scores = cross_val_score(clf, X, y, cv = k_folds)

print("Cross Validation Scores: ", scores)
print("Average CV Score: ", scores.mean())
print("Number of CV Scores used in Average: ", len(scores))
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB

modelos = {
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    'Random Forest': RandomForestClassifier(random_state=42),
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42),
    'SVM': SVC(random_state=42),
    'KNN': KNeighborsClassifier(),
    'Naive Bayes': GaussianNB()
}

 
from sklearn.model_selection import StratifiedKFold, cross_validate

scoring = {
    'accuracy': 'accuracy',
    'f1': 'f1',
    'recall': 'recall',
    'precision': 'precision'
}

k_folds = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)

resultados = {}

for nome, modelo in modelos.items():
    resultado_cv = cross_validate(modelo, X_resampled, y_resampled, cv=k_folds, scoring=scoring)
    resultados[nome] = resultado_cv
    print(f"\nModelo: {nome}")
    for metric in scoring.keys():
        scores = resultado_cv[f'test_{metric}']
        print(f"{metric.capitalize()}: Média = {scores.mean():.4f} | Desvio = {scores.std():.4f}")

```

## Árvore de Decisão

### Árvore com RandomizedSearchCV

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.tree import DecisionTreeClassifier, plot_tree
from scipy.stats import randint
from sklearn.model_selection import RandomizedSearchCV
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Parâmetros para busca aleatória
param_dist = {
    'criterion': ['gini', 'entropy'],
    'max_depth': randint(1, 50),
    'min_samples_split': randint(2, 20),
    'min_samples_leaf': randint(1, 10),
}

# Função para plotar a árvore
def plot_decision_tree(model, feature_names, class_names, title):
    plt.figure(figsize=(30, 15))
    plot_tree(model, filled=True, feature_names=feature_names, class_names=class_names, rounded=True, fontsize=10)
    plt.title(title)
    plt.show()

# RandomizedSearchCV com 100 iterações e 5-fold CV
random_search = RandomizedSearchCV(
    DecisionTreeClassifier(random_state=42),
    param_dist,
    n_iter=100,
    cv=5,
    scoring='accuracy',
    random_state=42
)

random_search.fit(X_treino, y_treino)

# Mostrar os melhores hiperparâmetros encontrados
print("\nMelhores parâmetros encontrados com RandomizedSearchCV:")
print(random_search.best_params_)
print("Melhor média de acurácia na validação cruzada: {:.2f}".format(random_search.best_score_))

```

```python
best_model_tree = DecisionTreeClassifier(
    criterion='gini',
    max_depth=3,
    min_samples_leaf=5,
    min_samples_split=8,
    random_state=42
)

best_model_tree.fit(X_treino, y_treino)

plot_decision_tree(
    best_model_tree,
    feature_names=X_resampled.columns,
    class_names=['Sem DC', 'Com DC'],
    title="Árvore de Decisão - RandomizedSearchCV"
)
```

```python
previsoes_tree = best_model_tree.predict(X_teste)
accuracy_tree = accuracy_score(y_teste, previsoes_tree)
print("Acurácia do melhor modelo (RandomizedSearchCV) no conjunto de teste: {:.2f}".format(accuracy_tree))

print(classification_report(y_teste, previsoes_tree))

from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(y_teste, previsoes_tree, display_labels=['Sem DC', 'Com DC'], cmap='Blues')
plt.title('Matriz de Confusão - Teste Holdout')
plt.show()
```

### Árvore com BayesSearch

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from skopt import BayesSearchCV
from skopt.space import Integer, Categorical

# Espaço de busca para BayesSearchCV
search_spaces = {
    'criterion': Categorical(['gini', 'entropy']),
    'max_depth': Integer(1, 50),
    'min_samples_split': Integer(2, 20),
    'min_samples_leaf': Integer(1, 10),
}

# Função para plotar a árvore
def plot_decision_tree(model, feature_names, class_names, title):
    plt.figure(figsize=(30, 15))
    plot_tree(model, filled=True, feature_names=feature_names, class_names=class_names, rounded=True, fontsize=10)
    plt.title(title)
    plt.show()

# BayesSearchCV com 100 iterações e 5-fold CV
bayes_search = BayesSearchCV(
    estimator=DecisionTreeClassifier(random_state=42),
    search_spaces=search_spaces,
    n_iter=100,
    cv=5,
    scoring='accuracy',
    random_state=42,
    n_jobs=-1
)

# Treinar nos dados balanceados (treino apenas!)
bayes_search.fit(X_treino, y_treino)

# Mostrar os melhores hiperparâmetros encontrados
print("\nMelhores parâmetros encontrados com BayesSearchCV:")
print(bayes_search.best_params_)
print("Melhor média de acurácia na validação cruzada: {:.2f}".format(bayes_search.best_score_))
```

```python
# Cria o modelo com os melhores parâmetros encontrados pelo BayesSearchCV
best_model_tree = DecisionTreeClassifier(
    **bayes_search.best_params_,
    random_state=42
)

# Treina o modelo com os dados balanceados completos
best_model_tree.fit(X_treino, y_treino)

# Plota a árvore
plot_decision_tree(
    best_model_tree,
    feature_names=X_resampled.columns,
    class_names=['Sem DC', 'Com DC'],
    title="Árvore de Decisão - BayesSearchCV"
)
```

```python
from sklearn.tree import export_text
import pandas as pd

# Previsões no treino para análise das folhas
previsoes = best_model_tree.predict(X_treino)

# Identifica a folha (nó terminal) de cada amostra
folhas = best_model_tree.apply(X_treino)

# Monta DataFrame com informações por amostra
df_folhas = pd.DataFrame({
    'folha': folhas,
    'verdadeiro': y_treino,
    'predito': previsoes
})

# Agrupa por folha e calcula acertos
estatisticas = df_folhas.groupby('folha').agg(
    total=('folha', 'count'),
    acertos=('verdadeiro', lambda x: (x == df_folhas.loc[x.index, 'predito']).sum())
).reset_index()

estatisticas['precisao'] = estatisticas['acertos'] / estatisticas['total']
estatisticas = estatisticas.sort_values(by='precisao', ascending=False)

# Exibe número de regras (folhas)
print(f"\nTotal de regras (folhas): {len(estatisticas)}")

# Exibe resumo por folha
print("\nResumo por folha (regra):")
print(estatisticas)

# Exporta as regras em texto
print("\nRegras da árvore de decisão:\n")
print(export_text(best_model_tree, feature_names=list(X_resampled.columns)))

# Regra com maior precisão
melhor_regra = estatisticas.iloc[0]

print("\nRegra com maior precisão:")
print(f"- Folha: {melhor_regra['folha']}")
print(f"- Precisão: {melhor_regra['precisao']:.2f}")
print(f"- Total de registros: {melhor_regra['total']}")
print(f"- Acertos: {melhor_regra['acertos']}")
```

```python
# Gerar previsões com o melhor modelo encontrado pela busca bayesiana
previsoes_tree = best_model_tree.predict(X_teste)

# Avaliar acurácia
accuracy_tree = accuracy_score(y_teste, previsoes_tree)
print("Acurácia do melhor modelo (BayesSearchCV) no conjunto de teste: {:.2f}".format(accuracy_tree))

# Relatório completo (precision, recall, f1-score)
print(classification_report(y_teste, previsoes_tree))

# Matriz de confusão
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(
    y_teste,
    previsoes_tree,
    display_labels=['Sem DC', 'Com DC'],
    cmap='Blues'
)
plt.title('Matriz de Confusão - Teste Holdout (BayesSearchCV)')
plt.show()
```

## Floresta Aleatória

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from skopt import BayesSearchCV
from skopt.space import Integer, Categorical
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Corrigir a definição de espaço de busca
param_bayes = {
    'criterion':  Categorical(['gini', 'entropy']),
    'max_depth': Integer(1, 50),
    'min_samples_split': Integer(2, 20),
    'min_samples_leaf': Integer(1, 10),
    'n_estimators': Integer(50, 120),
    'max_features': Integer(2, 5)
}

# Criar o otimizador bayesiano
bayes_search = BayesSearchCV(
    RandomForestClassifier(random_state=42),
    param_bayes,
    n_iter=50,
    cv=5,
    scoring='accuracy',
    random_state=42,
    n_jobs=-1
)

# Treinar no conjunto balanceado
bayes_search.fit(X_treino, y_treino)

# Mostrar os melhores parâmetros encontrados
print("\nMelhores parâmetros encontrados com BayesSearchCV:")
print(bayes_search.best_params_)
print("Melhor pontuação de validação cruzada: {:.2f}".format(bayes_search.best_score_))
```

```python
best_model_bayes = RandomForestClassifier(
    criterion='entropy',
    max_depth=50,
    max_features=3,
    min_samples_leaf=3,
    min_samples_split=2,
    n_estimators=120,
    random_state=42
)

# Corrigido aqui:
best_model_bayes.fit(X_treino, y_treino)

# Fazer previsões no conjunto de teste
previsoes_teste = best_model_bayes.predict(X_teste)

print(classification_report(y_teste, previsoes_teste))

cfm_matrix_teste = confusion_matrix(y_teste, previsoes_teste)

# Com mais clareza visual
sns.heatmap(cfm_matrix_teste, cmap='coolwarm', annot=True, linewidth=1, fmt='d',
            xticklabels=['Sem DC', 'Com DC'], yticklabels=['Sem DC', 'Com DC'])
plt.title("Matriz de Confusão - Teste")
plt.xlabel("Previsto")
plt.ylabel("Real")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Obter a importância das features do modelo treinado
importancias = best_model_bayes.feature_importances_

# Usar os nomes das features correspondentes ao treino balanceado
nomes_das_features = X_treino.columns  

# Ordenar índices das features da maior para menor importância
indices_ordenados = np.argsort(importancias)[::-1]

# Exibir as features mais importantes com proteção contra erro de índice
print("Variáveis mais relevantes:")
for idx in indices_ordenados:
    if idx < len(nomes_das_features):
        print(f"{nomes_das_features[idx]}: {importancias[idx]:.4f}")
    else:
        print(f"[AVISO] Índice {idx} fora do range dos nomes das features")

# Plotar a importância das features
plt.figure(figsize=(10, 6))
plt.title("Importância das Features")

plt.barh(range(len(importancias)), importancias[indices_ordenados], align="center")

# Converter nomes para array e proteger o ytick labels
nomes_array = np.array(nomes_das_features)
nomes_array_safe = nomes_array[indices_ordenados[:len(nomes_array)]]

plt.yticks(range(len(nomes_array_safe)), nomes_array_safe)
plt.gca().invert_yaxis()  # para as mais importantes ficarem no topo
plt.tight_layout()
plt.show()
```
